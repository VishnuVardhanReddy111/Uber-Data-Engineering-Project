```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.sql('select * from uber.bronze_new.stg_rides')
```

```python
df.display()
```

```python
df.select("driver_name").show()
```

```python
df.filter(col('vehicle_color')=='Red').display()
```

```python
df = df.where(col("vehicle_color")=="Red").display()
```

```python
df_tip = df.withColumn('tip_amount_plus_20', col('tip_amount') + 20)
df_tip = df_tip.select("tip_amount", "tip_amount_plus_20")
df_tip.display()
```

```python
df_renamed = df.withColumnRenamed("subtotal","Total")
```

```python
df_renamed.display()
```

```python
from pyspark.sql.functions import col
df.select('vehicle_color').distinct().display()
```

```python
df_new= df.groupBy("vehicle_color").agg(sum("subtotal"))
df_new.display()
```
