```python
from jinja2 import Template

jinja_str = '''
    select 
        {% for config in jinja_config %}
        {{config.select}}
            {% if not loop.last %}
            ,
            {% endif %}
        {% endfor %}
    from
        {% for config in jinja_config %}
                {% if loop.first %}
                    {{config.table}}
                {% else %}
                    left join {{config.table}} on {{config.on}}
                {% endif %}
        {% endfor %}

        {% for config in jinja_config %}
            {% if loop.first %}
                {% if config.where != "" %}
                    where
                {% endif %}
            {% endif %}
            {{config.where}}
                {% if not loop.last %}
                    {%if config.where != "" %}
                        and
                    {% endif %}
                {% endif %}
        {% endfor %}
'''

template = Template(jinja_str)
rendered_template = template.render(jinja_config=jinja_config)
display(spark.sql(rendered_template))
```
